```python
import opendatasets as od
import os

NOWA_NAZWA = "data"
STARA_NAZWA = "book-recommendation-dataset"

# Podajesz bezpośredni link do strony z datasetem
od.download("https://www.kaggle.com/datasets/arashnic/book-recommendation-dataset/data")

# zmiana nazwy folderu na "data"
if os.path.exists(STARA_NAZWA):
    # Sprawdzamy na wypadek, gdyby folder o nowej nazwie już istniał
    if not os.path.exists(NOWA_NAZWA):
        os.rename(STARA_NAZWA, NOWA_NAZWA)
        print(f"Sukces! Folder został pobrany i przemianowany na: '{NOWA_NAZWA}'")
    else:
        print(f"Folder o nazwie '{NOWA_NAZWA}' już istnieje w tym katalogu!")
```
